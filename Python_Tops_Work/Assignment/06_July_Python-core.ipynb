GENERATOR
no need for class
it is way  to create an iterator using YIELD
-state saved automatically
- faster

```python
def count_value(n):
    for i in range(1,n+1):
        yield i
```

```python
gen = count_value(10)

print(next(gen))
print(next(gen))
print(next(gen))
```

```python
class Employee :
    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self) :
        return self.__salary
    
    def set_salary(self, amount) :
        if amount > 0 :
            self.__salary = amount
```

```python
e = Employee(200)
print(e.get_salary())

e2 = Employee(200)
e2.set_salary(250)
print(e2.get_salary())
```

ABSTRACTION

hiding internal implementation and exposing only what is necessary

Abstract Base Classes
ABC

```python
from abc import ABC, abstractmethod

class DataSource(ABC) :
    
    @abstractmethod
    def read_data(self):
        pass
```

```python
class CSVSource(DataSource) :
    def read_data(self) :
        return "reading file"
```

```python
cs1 = CSVSource()

kk = cs1.read_data()

print(kk)
```
